```python
import pandas as pd
file_path = "C:\Devdan College\DCU work\Statisticsal Data Analytics\open_food_facts_2025 (1).csv"
df = pd.read_csv(file_path)

df.head()
```

```python
print("Duplicates found:", df.duplicated().sum()) #Finding Dupilcates in the selected dataset
```

```python
# Selecting the assigned columns for data cleaning
target_cols = [
    'image_url', 'creator', 'created_date', 'last_modified_date', 'completeness', 
    'nutriscore_grade', 'ingredients_text', 'allergens', 'additives_n', 'additives_en'
]

# Drops the row if ANY of the 10 target columns contain a null value
df = df.dropna(subset=target_cols)

# Verify how many nulls remain in your 10 columns
print(df[target_cols].isnull().sum())

df = df.drop_duplicates()
```

```python
df['created_date'] = pd.to_datetime(df['created_date'], errors='coerce') #Changing the date format to default format pandas accept
df['last_modified_date'] = pd.to_datetime(df['last_modified_date'], errors='coerce')#Changing the date format to default format pandas accept
```

```python
# Check the data types of your 10 columns
print(df[target_cols].dtypes)

# Preview just the 10 cleaned columns
df[target_cols].head()
```

```python
# Remove the "en:" prefix from all rows in the allergens column
df['allergens'] = df['allergens'].str.replace('en:', '')

#Replace the commas with a comma and a space for better readability
df['allergens'] = df['allergens'].str.replace(',', ', ')

# Check the data types of your 10 columns
print(df[target_cols].dtypes)

# Preview just the 10 cleaned columns
df[target_cols].head()
```

```python
# Save only the target columns as a CSV
df[target_cols].to_csv('cleaned_10_columns.csv', index=False)
```

```python
# 1. Fix Allergen Spacing
# Using regex=True ensures that any weird multiple spaces after a comma become exactly one space
df['allergens'] = df['allergens'].str.replace(r',\s+', ', ', regex=True)

# 2. Convert Date Columns to Proper Datetime Format
df['created_date'] = pd.to_datetime(df['created_date'])
df['last_modified_date'] = pd.to_datetime(df['last_modified_date'])

# 3. Convert Additives Count from Decimal (float) to Integer (int)
df['additives_n'] = df['additives_n'].astype(int)

# 4. Save the fully polished dataset!
df.to_csv('fully_cleaned_10_columns.csv', index=False)
```

```python
# 1. Remove unwanted symbols
# In Open Food Facts datasets, allergens in the ingredients text are often 
# surrounded by underscores (e.g., "_Lait_"). This removes them.
#df['ingredients_text'] = df['ingredients_text'].str.replace('_', '', regex=False)

# You can add additional characters to remove if you spot them, like asterisks:
df['ingredients_text'] = df['ingredients_text'].str.replace('*', '', regex=False)

# 2. Format to Sentence Case (Uppercase first letter, lowercase the rest)
df['ingredients_text'] = df['ingredients_text'].str.capitalize()

# 3. Define the 10 columns you want to keep
columns_to_keep = [
    'image_url', 'creator', 'created_date', 'last_modified_date', 
    'completeness', 'nutriscore_grade', 'ingredients_text', 
    'allergens', 'additives_n', 'additives_en'
]

# 4. Filter the dataset to include only those columns, then save it
df[columns_to_keep].to_csv('final_10_columns_only.csv', index=False)
```
