**DATA CLEANING**


Step 1 — Create a safe working copy

```python
import pandas as pd
import numpy as np
import re

df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy:", df_clean.shape)
```

Step 2 — Standardise empty values and whitespace

```python
# Remove leading/trailing whitespace and extra spaces
for col in df_clean.columns:
    df_clean[col] = (
        df_clean[col]
        .str.strip()
        .str.replace(r'\s+', ' ', regex=True)
    )

# Replace common missing-value placeholders with actual nulls
missing_tokens = [
    '', 'null', 'none', 'nan', 'n/a', 'na',
    'not available', 'not provided'
]

for col in df_clean.columns:
    mask = df_clean[col].str.lower().isin(missing_tokens)
    df_clean.loc[mask.fillna(False), col] = pd.NA

print("Text formatting completed.")
```

Step 3 — Clean product codes

```python
# Remove internal spaces from codes
df_clean['code'] = (
    df_clean['code']
    .str.replace(r'\s+', '', regex=True)
)

# Check codes that contain non-digit characters
invalid_codes = (
    df_clean['code'].notna() &
    ~df_clean['code'].str.fullmatch(r'\d+').fillna(False)
)

print("Non-numeric product codes:", invalid_codes.sum())

display(df_clean.loc[
    invalid_codes, ['code', 'product_name']
].head(10))
```

  Step 4 — Validate URLs

```python
valid_urls = df_clean['url'].str.match(
    r'^https?://\S+$',
    na=False
)

invalid_urls = df_clean['url'].notna() & ~valid_urls

print("Invalid URL formats:", invalid_urls.sum())

display(df_clean.loc[
    invalid_urls, ['code', 'url']
].head(10))
```

Step 5 — Clean product and generic names

```python
for col in ['product_name', 'generic_name']:
    df_clean[col] = (
        df_clean[col]
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )

print("Missing product names:",
      df_clean['product_name'].isna().sum())

print("Missing generic names:",
      df_clean['generic_name'].isna().sum())
```

Step 6 — Standardise quantity formatting

```python
df_clean['quantity'] = (
    df_clean['quantity']
    .str.replace(',', '.', regex=False)
    .str.replace(
        r'(?i)(\d)\s*(kg|mg|ml|cl|g|l)\b',
        r'\1 \2',
        regex=True
    )
    .str.strip()
)

print("Top 15 quantities after cleaning:")
display(df_clean['quantity'].value_counts().head(15))
```

Step 7 — Clean food and PNNS categories

```python
category_cols = [
    'food_groups_en',
    'pnns_groups_1',
    'pnns_groups_2'
]

for col in category_cols:
    df_clean[col] = (
        df_clean[col]
        .str.replace(r'\s*,\s*', ', ', regex=True)
        .str.strip()
    )

for col in category_cols:
    print(f"\n{col}:")
    print("Unique values:", df_clean[col].nunique())
    display(df_clean[col].value_counts().head(10))
```

Step 8 — Clean NOVA group values

```python
nova_numeric = pd.to_numeric(
    df_clean['nova_group'],
    errors='coerce'
)

invalid_nova = (
    df_clean['nova_group'].notna()
    & ~nova_numeric.isin([1, 2, 3, 4])
)

print("Invalid NOVA values:", invalid_nova.sum())

# Keep only valid classifications, others become missing
df_clean['nova_group'] = (
    nova_numeric
    .where(nova_numeric.isin([1, 2, 3, 4]))
    .astype('Int64')
)

display(
    df_clean['nova_group'].value_counts(
        dropna=False
    ).sort_index()
)
```

Step 9 — Validate NOVA group labels

```python
display(
    pd.crosstab(
        df_clean['nova_group'],
        df_clean['nova_group_label'],
        dropna=False
    )
)
```

Step 10 — Remove exact duplicate rows

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Exact duplicates removed:", before - after)
```

```python
df_clean['quantity_missing'] = (
    df_clean['quantity'].isna().astype('int8')
)

display(
    df_clean[['product_name', 'quantity', 'quantity_missing']]
    .head(10)
)
```

```python
print("Top 30 quantity values:")
display(df_clean['quantity'].value_counts().head(30))

print("Missing quantities:")
print(df_clean['quantity'].isna().sum())
```

```python
# Extract simple quantities only
extracted = df_clean['quantity'].str.extract(
    r'(?i)^\s*(\d+(?:[.,]\d+)?)\s*(kg|g|mg|l|ml|cl)\s*$'
)

df_clean['quantity_value'] = pd.to_numeric(
    extracted[0].str.replace(',', '.', regex=False),
    errors='coerce'
)

df_clean['quantity_unit'] = extracted[1].str.lower()

display(
    df_clean[
        ['quantity', 'quantity_value', 'quantity_unit']
    ].head(15)
)
```

```python
conversion = {
    'kg': 1000,
    'g': 1,
    'mg': 0.001,
    'l': 1000,
    'ml': 1,
    'cl': 10
}

df_clean['quantity_standardised'] = (
    df_clean['quantity_value'] *
    df_clean['quantity_unit'].map(conversion)
)

df_clean['quantity_standard_unit'] = (
    df_clean['quantity_unit'].map({
        'kg': 'g',
        'g': 'g',
        'mg': 'g',
        'l': 'ml',
        'ml': 'ml',
        'cl': 'ml'
    })
)

display(
    df_clean[
        [
            'quantity',
            'quantity_standardised',
            'quantity_standard_unit'
        ]
    ].head(15)
)
```

```python
print("Total rows:", len(df_clean))

print(
    "Original quantity missing:",
    df_clean['quantity'].isna().sum()
)

print(
    "Simple quantities successfully parsed:",
    df_clean['quantity_value'].notna().sum()
)

print(
    "Non-missing quantities not parsed:",
    (
        df_clean['quantity'].notna()
        & df_clean['quantity_value'].isna()
    ).sum()
)

print(
    "Quantity missing percentage:",
    round(df_clean['quantity'].isna().mean() * 100, 2)
)
```
